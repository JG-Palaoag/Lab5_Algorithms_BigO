```python
# [CELL 1] STUDENT CONFIGURATION

import time
import random
import statistics

LAST_NAME = "Palaoag"            
STUDENT_ID = "TUPM-26-1949"       
FAVORITE_ARTIST = "Rico Blanco"  

id_digits = [int(c) for c in STUDENT_ID if c.isdigit()]
name_letters = [c for c in LAST_NAME.upper() if c.isalpha()]
artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]

SEED_DIGIT = id_digits[-1]
ID_SUM = sum(id_digits)
NAME_LENGTH = len(name_letters) 
ARTIST_LENGTH = len(artist_letters) 
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in artist_letters)

print(f"LAST_NAME       = {LAST_NAME}")
print(f"STUDENT_ID      = {STUDENT_ID}")
print(f"FAVORITE_ARTIST = {FAVORITE_ARTIST}")
print(f"SEED_DIGIT    = {SEED_DIGIT}")
print(f"ID_SUM        = {ID_SUM}")
print(f"NAME_LENGTH   = {NAME_LENGTH}")
print(f"ARTIST_LENGTH = {ARTIST_LENGTH}")
print(f"ARTIST_SUM    = {ARTIST_SUM}")
```

```python
# [CELL 2] LARGEST VALUE

def measure(func, *args, repeats=5):
    """Python runtime test: average wall-clock time (seconds) over `repeats` runs."""
    times = []
    result = None
    for _ in range(repeats):
        start = time.perf_counter()
        result = func(*args)
        times.append(time.perf_counter() - start)
    return result, statistics.mean(times)


def largest_value(values):
    """Largest Value Algorithm. Returns (max, comparisons). O(n) time, O(1) space."""
    largest = values[0]
    comparisons = 0
    for v in values[1:]:
        comparisons += 1
        if v > largest:
            largest = v
    return largest, comparisons

print(f"{'n':>10} {'comparisons':>12} {'avg time (ms)':>15}")
for n in [1_000, 2_000, 4_000, 8_000, 16_000, 32_000]:
    data = [(ID_SUM * 31 + i * 17) % 100_003 for i in range(n)]
    (largest, comps), t = measure(largest_value, data, repeats=10)
    print(f"{n:>10} {comps:>12} {t * 1000:>15.4f}")

print("\nObservation: comparisons = n - 1, so growth is LINEAR -> O(n).")
```

```python
# [CELL 3] BIG-O TABLE

big_o_table = [
    ("Largest Value Algorithm", "O(n)", "O(1)"),
    ("Linear Search", "O(n)", "O(1)"),
    ("Binary Search (rotated list)", "O(log n)", "O(1)"),
    ("Merge two sorted lists", "O(m + n)", "O(m + n)"),
    ("Bubble Sort", "O(n^2)", "O(1)"),
    ("Selection Sort", "O(n^2)", "O(1)"),
    ("Insertion Sort", "O(n^2)", "O(1)"),
]
print(f"{'Algorithm':<30}{'Time':<12}{'Space':<10}")
for name, t, s in big_o_table:
    print(f"{name:<30}{t:<12}{s:<10}")
```

```python
# [CELL 4]

SIZE_A = 10 + SEED_DIGIT
STEP_A = (SEED_DIGIT % 4) + 2
dataset_a = [ID_SUM + i * STEP_A for i in range(SIZE_A)]

SIZE_B = ARTIST_LENGTH + 5
STEP_B = (NAME_LENGTH % 4) + 2
dataset_b = [NAME_LENGTH + ARTIST_LENGTH + i * STEP_B for i in range(SIZE_B)] 

print(f"Department A dataset ({len(dataset_a)}): {dataset_a}")
print(f"Department B dataset ({len(dataset_b)}): {dataset_b}")
```

```python
# [CELL 5]

def merge_sorted(a, b):
    """Two-pointer merge. Returns (merged, comparisons). O(m+n) time, O(m+n) space."""
    i = j = 0
    merged = []
    comparisons = 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    merged.extend(a[i:]) 
    merged.extend(b[j:])
    return merged, comparisons


(consolidated, merge_comps), merge_time = measure(merge_sorted, dataset_a, dataset_b, repeats=1000)

is_ordered = all(consolidated[k] <= consolidated[k + 1] for k in range(len(consolidated) - 1))
count_ok = len(consolidated) == len(dataset_a) + len(dataset_b)
matches_builtin = consolidated == sorted(dataset_a + dataset_b)

print("\n=== EXERCISE 2 RESULTS ===")
print(f"Consolidated dataset : {consolidated}")
print(f"Records before       : A={len(dataset_a)}, B={len(dataset_b)}, total={len(dataset_a) + len(dataset_b)}")
print(f"Records after        : {len(consolidated)}")
print(f"Comparison count     : {merge_comps}")
print(f"Merge time           : {merge_time * 1e6:.2f} us")
print(f"Ordering verified    : {is_ordered}")
print(f"Record count verified: {count_ok}")
print(f"Matches sorted()     : {matches_builtin}")
print("Time complexity : O(m + n)   (comparisons <= m + n - 1)")
print("Space complexity: O(m + n)   (new consolidated list)")
```

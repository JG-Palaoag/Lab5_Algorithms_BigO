```python
# [CELL 1] STUDENT CONFIGURATION

import time
import random
import statistics

LAST_NAME = "Palaoag"            
STUDENT_ID = "TUPM-26-1949"       
FAVORITE_ARTIST = "Rico Blanco"  

id_digits = [int(c) for c in STUDENT_ID if c.isdigit()]
name_letters = [c for c in LAST_NAME.upper() if c.isalpha()]
artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]

SEED_DIGIT = id_digits[-1]
ID_SUM = sum(id_digits)
NAME_LENGTH = len(name_letters) 
ARTIST_LENGTH = len(artist_letters) 
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in artist_letters)

print(f"LAST_NAME       = {LAST_NAME}")
print(f"STUDENT_ID      = {STUDENT_ID}")
print(f"FAVORITE_ARTIST = {FAVORITE_ARTIST}")
print(f"SEED_DIGIT    = {SEED_DIGIT}")
print(f"ID_SUM        = {ID_SUM}")
print(f"NAME_LENGTH   = {NAME_LENGTH}")
print(f"ARTIST_LENGTH = {ARTIST_LENGTH}")
print(f"ARTIST_SUM    = {ARTIST_SUM}")
```

```python
# [CELL 2] LARGEST VALUE

def measure(func, *args, repeats=5):
    """Python runtime test: average wall-clock time (seconds) over `repeats` runs."""
    times = []
    result = None
    for _ in range(repeats):
        start = time.perf_counter()
        result = func(*args)
        times.append(time.perf_counter() - start)
    return result, statistics.mean(times)


def largest_value(values):
    """Largest Value Algorithm. Returns (max, comparisons). O(n) time, O(1) space."""
    largest = values[0]
    comparisons = 0
    for v in values[1:]:
        comparisons += 1
        if v > largest:
            largest = v
    return largest, comparisons

print(f"{'n':>10} {'comparisons':>12} {'avg time (ms)':>15}")
for n in [1_000, 2_000, 4_000, 8_000, 16_000, 32_000]:
    data = [(ID_SUM * 31 + i * 17) % 100_003 for i in range(n)]
    (largest, comps), t = measure(largest_value, data, repeats=10)
    print(f"{n:>10} {comps:>12} {t * 1000:>15.4f}")

print("\nObservation: comparisons = n - 1, so growth is LINEAR -> O(n).")
```

```python
# [CELL 3] BIG-O TABLE

big_o_table = [
    ("Largest Value Algorithm", "O(n)", "O(1)"),
    ("Linear Search", "O(n)", "O(1)"),
    ("Binary Search (rotated list)", "O(log n)", "O(1)"),
    ("Merge two sorted lists", "O(m + n)", "O(m + n)"),
    ("Bubble Sort", "O(n^2)", "O(1)"),
    ("Selection Sort", "O(n^2)", "O(1)"),
    ("Insertion Sort", "O(n^2)", "O(1)"),
]
print(f"{'Algorithm':<30}{'Time':<12}{'Space':<10}")
for name, t, s in big_o_table:
    print(f"{name:<30}{t:<12}{s:<10}")
```

```python
# [CELL 4] ROTATED RECORDS

N1 = 20 + NAME_LENGTH + ARTIST_LENGTH
STEP1 = (SEED_DIGIT % 5) + 3
START1 = 1000 + ID_SUM * 10

original_records = [START1 + i * STEP1 for i in range(N1)] 
rotation_point = (SEED_DIGIT % (N1 - 1)) + 1 
rotated_records = original_records[rotation_point:] + original_records[:rotation_point]

target_index = (ID_SUM + ARTIST_SUM) % N1
target_incident = original_records[target_index] 
missing_incident = target_incident + 1 

print(f"Number of records : {N1}")
print(f"Rotation point    : {rotation_point}")
print(f"Original records  : {original_records}")
print(f"Rotated records   : {rotated_records}")
print(f"Target incident   : {target_incident}")
print(f"Missing test value: {missing_incident}")
```

```python
# [CELL 5] ROTATED BINARY SEARCH

def linear_search(arr, target):
    """Returns (index or -1, comparisons). O(n) time, O(1) space."""
    comparisons = 0
    for i, v in enumerate(arr):
        comparisons += 1
        if v == target:
            return i, comparisons
    return -1, comparisons


def rotated_binary_search(arr, target):
    """
    Binary search on a rotated sorted list. O(log n) time, O(1) space.
    Comparison counting: each decision (equality test, which-half-is-sorted
    test, target-range test) counts as 1 comparison.
    """
    lo, hi = 0, len(arr) - 1
    comparisons = 0
    while lo <= hi:
        mid = (lo + hi) // 2
        comparisons += 1
        if arr[mid] == target:
            return mid, comparisons
        comparisons += 1 
        if arr[lo] <= arr[mid]: 
            comparisons += 1  
            if arr[lo] <= target < arr[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:  
            comparisons += 1  
            if arr[mid] < target <= arr[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1, comparisons
```

```python
# [CELL 6] FINAL OUTPUT

def report(label, target):
    (b_idx, b_comps), b_time = measure(rotated_binary_search, rotated_records, target, repeats=1000)
    (l_idx, l_comps), l_time = measure(linear_search, rotated_records, target, repeats=1000)
    print(f"--- {label}: target = {target} ---")
    for name, idx, comps, t in [("Binary Search", b_idx, b_comps, b_time),
                                ("Linear Search", l_idx, l_comps, l_time)]:
        pos = f"found at index {idx}" if idx != -1 else "NOT FOUND"
        print(f"{name:<14}: {pos:<20} comparisons = {comps:<4} time = {t * 1e6:.2f} us")
    return b_idx, b_comps, l_idx, l_comps


ex1_found = report("Target incident", target_incident)
ex1_missing = report("Missing incident", missing_incident)

print("\n===FINAL OUTPUT===")
print(f"Target {target_incident} -> position {ex1_found[0]} in rotated list "
      f"(original index {target_index}).")
print(f"Binary comparisons: {ex1_found[1]} | Linear comparisons: {ex1_found[3]}")
print("Time complexity : Binary O(log n) vs Linear O(n)")
print("Space complexity: O(1) for both (iterative, no extra structures)")
```

```python
# [CELL 1] STUDENT CONFIGURATION

import time
import random
import statistics

LAST_NAME = "Palaoag"            
STUDENT_ID = "TUPM-26-1949"       
FAVORITE_ARTIST = "Rico Blanco"  

id_digits = [int(c) for c in STUDENT_ID if c.isdigit()]
name_letters = [c for c in LAST_NAME.upper() if c.isalpha()]
artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]

SEED_DIGIT = id_digits[-1]
ID_SUM = sum(id_digits)
NAME_LENGTH = len(name_letters) 
ARTIST_LENGTH = len(artist_letters) 
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in artist_letters)

print(f"LAST_NAME       = {LAST_NAME}")
print(f"STUDENT_ID      = {STUDENT_ID}")
print(f"FAVORITE_ARTIST = {FAVORITE_ARTIST}")
print(f"SEED_DIGIT    = {SEED_DIGIT}")
print(f"ID_SUM        = {ID_SUM}")
print(f"NAME_LENGTH   = {NAME_LENGTH}")
print(f"ARTIST_LENGTH = {ARTIST_LENGTH}")
print(f"ARTIST_SUM    = {ARTIST_SUM}")
```

```python
# [CELL 2] LARGEST VALUE

def measure(func, *args, repeats=5):
    """Python runtime test: average wall-clock time (seconds) over `repeats` runs."""
    times = []
    result = None
    for _ in range(repeats):
        start = time.perf_counter()
        result = func(*args)
        times.append(time.perf_counter() - start)
    return result, statistics.mean(times)


def largest_value(values):
    """Largest Value Algorithm. Returns (max, comparisons). O(n) time, O(1) space."""
    largest = values[0]
    comparisons = 0
    for v in values[1:]:
        comparisons += 1
        if v > largest:
            largest = v
    return largest, comparisons

print(f"{'n':>10} {'comparisons':>12} {'avg time (ms)':>15}")
for n in [1_000, 2_000, 4_000, 8_000, 16_000, 32_000]:
    data = [(ID_SUM * 31 + i * 17) % 100_003 for i in range(n)]
    (largest, comps), t = measure(largest_value, data, repeats=10)
    print(f"{n:>10} {comps:>12} {t * 1000:>15.4f}")

print("\nObservation: comparisons = n - 1, so growth is LINEAR -> O(n).")
```

```python
# [CELL 3] BIG-O TABLE

big_o_table = [
    ("Largest Value Algorithm", "O(n)", "O(1)"),
    ("Linear Search", "O(n)", "O(1)"),
    ("Binary Search (rotated list)", "O(log n)", "O(1)"),
    ("Merge two sorted lists", "O(m + n)", "O(m + n)"),
    ("Bubble Sort", "O(n^2)", "O(1)"),
    ("Selection Sort", "O(n^2)", "O(1)"),
    ("Insertion Sort", "O(n^2)", "O(1)"),
]
print(f"{'Algorithm':<30}{'Time':<12}{'Space':<10}")
for name, t, s in big_o_table:
    print(f"{name:<30}{t:<12}{s:<10}")
```

```python
# [CELL 4] ORDER NUMBERS

N3 = 100 + ID_SUM
SEED3 = ID_SUM * 100 + ARTIST_SUM
rng = random.Random(SEED3)

order_numbers = rng.sample(range(10_000, 99_999), N3)

inputs = {
    "Sorted": sorted(order_numbers),
    "Reverse": sorted(order_numbers, reverse=True),
    "Random": order_numbers[:],
}

print(f"Number of orders: {N3}   (seed = {SEED3})")
for name, data in inputs.items():
    print(f"{name:<8} first 10: {data[:10]}")
assert sorted(inputs["Random"]) == inputs["Sorted"] == inputs["Reverse"][::-1]
print("All three conditions contain the same set of orders: True")
```

```python
# [CELL 5] ALGORITHMS

def bubble_sort(data):
    """Bubble Sort with early exit. Movement = one swap."""
    a = data[:]
    comps = moves = 0
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            comps += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                moves += 1
                swapped = True
        if not swapped:
            break
    return a, comps, moves


def selection_sort(data):
    """Selection Sort. Movement = one swap (only when the minimum is not in place)."""
    a = data[:]
    comps = moves = 0
    n = len(a)
    for i in range(n - 1):
        m = i
        for j in range(i + 1, n):
            comps += 1
            if a[j] < a[m]:
                m = j
        if m != i:
            a[i], a[m] = a[m], a[i]
            moves += 1
    return a, comps, moves


def insertion_sort(data):
    """Insertion Sort. Movement = each element shift + the final placement of the key."""
    a = data[:]
    comps = moves = 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comps += 1
            if a[j] > key:
                a[j + 1] = a[j]
                moves += 1
                j -= 1
            else:
                break
        if j + 1 != i:
            a[j + 1] = key
            moves += 1
    return a, comps, moves


algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort,
}
```

```python
# [CELL 6] EXECUTION-TIME MEASUREMENT

def is_sorted(a):
    return all(a[k] <= a[k + 1] for k in range(len(a) - 1))


REPEATS = 20
results = []

for cond, data in inputs.items():
    for alg_name, alg in algorithms.items():
        times = []
        for _ in range(REPEATS):
            start = time.perf_counter()
            out, comps, moves = alg(data)
            times.append(time.perf_counter() - start)
        ok = is_sorted(out) and out == sorted(data)
        results.append((cond, alg_name, comps, moves, statistics.mean(times) * 1000, ok))

print(f"{'Condition':<10}{'Algorithm':<16}{'Comparisons':>12}{'Movements':>11}{'Time (ms)':>12}{'Sorted?':>9}")
print("-" * 69)
for cond, alg_name, comps, moves, ms, ok in results:
    print(f"{cond:<10}{alg_name:<16}{comps:>12}{moves:>11}{ms:>12.4f}{str(ok):>9}")

print("\nAll results correctly sorted:", all(r[5] for r in results))
```

```python
# [CELL 7]

try:
    import matplotlib.pyplot as plt
    conds = list(inputs)
    algs = list(algorithms)
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    width = 0.25
    for k, alg in enumerate(algs):
        comps = [next(r[2] for r in results if r[0] == c and r[1] == alg) for c in conds]
        times = [next(r[4] for r in results if r[0] == c and r[1] == alg) for c in conds]
        xs = [i + k * width for i in range(len(conds))]
        axes[0].bar(xs, comps, width, label=alg)
        axes[1].bar(xs, times, width, label=alg)
    for ax, title in zip(axes, ["Comparisons", "Average time (ms)"]):
        ax.set_xticks([i + width for i in range(len(conds))])
        ax.set_xticklabels(conds)
        ax.set_title(title)
        ax.legend()
    plt.tight_layout()
    plt.show()
except ImportError:
    print("matplotlib not installed - run `pip install matplotlib` to see the charts.")
```

```python
# [CELL 8] COMPARISON

n = N3
print(f"n = {n}:  n(n-1)/2 = {n * (n - 1) // 2}\n")

print("FASTEST ALGORITHM")
for cond in inputs:
    best = min((r for r in results if r[0] == cond), key=lambda r: r[4])
    print(f"{cond:<8}: {best[1]}  ({best[4]:.4f} ms)")

print("""
BIG O AND CASE ANALYSIS
Algorithm       Best case        Average case   Worst case     Space
Bubble Sort     O(n)  sorted*    O(n^2)         O(n^2) reverse O(1)
Selection Sort  O(n^2) always    O(n^2)         O(n^2)         O(1)
Insertion Sort  O(n)  sorted     O(n^2)         O(n^2) reverse O(1)
(*Bubble Sort's O(n) best case relies on the early-exit flag used above.)

Comparison:
- Sorted input : Bubble and Insertion need ~n-1 comparisons (best case, linear);
                 Selection still performs n(n-1)/2 comparisons, but almost no swaps.
- Reverse input: every pair is out of order -> worst case O(n^2) for all three;
                 Bubble and Insertion perform n(n-1)/2 movements.
- Random input : average case O(n^2); Insertion Sort needs about half the
                 comparisons of Bubble/Selection.
- Selection Sort makes at most n-1 swaps (fewest movements), so in Python it can
                 win on reverse/random data even with more comparisons.
- Compare counts (machine-independent) with times (machine-dependent): the two
                 can rank the algorithms differently.
""")
```
